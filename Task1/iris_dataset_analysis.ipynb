# Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully")
```

```python
# since we know they are two way to load iris dataset 
# 1. By downloading iris csv file mannually and then load into my code.
# 2. By using built-in method  load_dataset() which provide seaborn library Now i will implement it by second method.
```

# 2. Load Dataset

```python
# Using seaborn built-in dataset
iris = sns.load_dataset('iris')
iris
# The other method you can load dataset by pd.read_csv("path of file") but first install csv file and then load csv by pd.read_csv method
# There are 150 rows/records in this iris dataset
```

# 3. Basic Data Inspection

```python
# .shape is an attribute to find the dimensions(means in simple words how many rows and columns) of an dataset
print("Shape of dataset:", iris.shape)
```

```python
# .columns is also an attribute that give columns info means all columns names of that dataset what datatype of that
print("Column Names:", iris.columns)
```

```python
# Now tell me what does head() function do it print/show first five rows/records of particular dataset
print("First 5 Rows:")
print(iris.head())
```

# 4. Dataset Information

```python
# info() is a method that give concise summary or a quick "health check" of the dataset
print("Dataset Info:")
iris.info()
```

```python
# describe() is a method that give statistical summary of a dataset means mean, standard deviation minimum or maximum etc and other values
# it help to understand dataset.
print("Statistical Summary:")
print(iris.describe())
```

# 5. Check for Missing Values



```python
#isnull().sum() is a method that give count of missing values that is available in each column
print("\nMissing Values:")
print(iris.isnull().sum())
```

# 6. Data Visualization

# 6.1 Scatter Plot

```python
# scatter plot basically we use visualizing the relationship, correlation, and patterns between two numerical variables.
# here we take two numerical variables sepal_lenght as x axis and sepal_width as y_axis
plt.figure(figsize=(8,6))
sns.scatterplot(x='sepal_length', y='sepal_width', hue='species', data=iris)
plt.title("Sepal Length vs Sepal Width")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='petal_length', y='petal_width', hue='species', data=iris)
plt.title("Petal Length vs Petal Width")
plt.show()
```

# 6.2 Histograms

```python
# Let's look at the distribution of each feature. 
# This helps us see if the data is balanced or if any features have weird outliers
#it creates a grid of histograms for every numerical column
iris.hist(figsize=(10,8))
plt.suptitle("Feature Distributions") # This adds a "Super Title" at the very top of the entire grid
plt.show() #This cleans up the output and displays the final figure
```

#  6.3 Box Plots

```python
# Now take a look on box plot.Its main job is to help you spot outliers
plt.figure(figsize=(10,6))
sns.boxplot(data=iris)
plt.title("Box Plot for Outlier Detection")
plt.show()
#This box plot reveals that while most features are well-distributed, sepal_width contains a few outliers that sit outside the expected range.
```

# 6.4 Pair Plot (All Relationships)
"""
A pair plot (often called a scatterplot matrix) is a data visualization tool that 
shows the relationships between all numerical variables in a dataset at once
"""

```python
# This gives us a massive overview of how all features interact. 
# It's the best way to see which combinations of traits make the species stand out from each other.
sns.pairplot(iris, hue='species')
plt.show()
```
